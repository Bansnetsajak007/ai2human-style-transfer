```python
import pandas as pd

# Reload original dataset
dataset_path = "AI_Human.csv"
df = pd.read_csv(dataset_path)

# Filter human text only
human_df = df[df['generated'] == 0.0].reset_index(drop=True)

# Sample 10k full paragraphs
sampled_df = human_df.sample(n=10000, random_state=42)

# Save properly - use separator between paragraphs
# Each paragraph on ONE line - replace internal newlines
with open("FINAL_10k.txt", "w", encoding="utf-8") as f:
    for text in sampled_df['text']:
        # Replace internal newlines with space
        clean_text = str(text).replace('\n', ' ').replace('\r', ' ').strip()
        f.write(clean_text + "\n\n\n")  # Double newline to separate paragraphs

# Verify
with open("FINAL_10k.txt", "r", encoding="utf-8", errors="replace") as f:
    lines = f.readlines()
print(f"Total lines: {len(lines)}")  # Should be exactly 10,000
print(f"Sample: {lines[0][:150]}")
```
